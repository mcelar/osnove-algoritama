## 3 Što ispisuje?

### Zadatak 7.

Što ispisuje program za n=12?

Što program radi?

**Odgovor:**

````python
n = 12
k = 0
for i in range(1, n + 1):
    if n % i == 0 and i % 2 == 0:
        k += 1
print(k)
````
